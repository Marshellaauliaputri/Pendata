### Data Preparation (Persiapan Data)

Tahap **persiapan data** merupakan fondasi penting dalam proses pengolahan data dan pembangunan model. Pada tahap ini, data yang telah dikumpulkan dari berbagai sumber akan diperiksa dan diproses agar siap digunakan dalam pemodelan. Kegiatan utama meliputi:

1. **Pemilihan Data yang Relevan**  
   Memastikan hanya data yang memiliki nilai informasi dan relevansi terhadap tujuan analisis yang dipilih. Hal ini membantu mengurangi kebisingan (noise) dan meningkatkan kualitas model.

2. **Pembersihan Data**  
   Mengatasi masalah seperti data yang hilang (missing values), data duplikat, atau data yang tidak konsisten. Pembersihan data ini penting agar model tidak terpengaruh oleh informasi yang salah atau tidak lengkap.

3. **Transformasi Data**  
   Melakukan normalisasi, standarisasi, encoding kategori, atau transformasi lainnya agar data sesuai dengan format yang dibutuhkan oleh algoritma. Misalnya, mengubah data teks menjadi vektor numerik untuk pemodelan machine learning.

4. **Pengaturan Format dan Struktur**  
   Menyesuaikan struktur data seperti kolom, tipe data, dan format tanggal agar seragam dan dapat diproses dengan lancar oleh perangkat lunak analisis atau algoritma.

Tahap ini memastikan data yang digunakan berkualitas tinggi, sehingga model yang dibangun nantinya lebih akurat dan dapat diandalkan.
