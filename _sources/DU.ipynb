## Data Understanding (Pemahaman Data)

merupakan tahap pengumpulan dan pengenalan data. Data dianalisis untuk mengetahui struktur, isi, dan kualitasnya melalui eksplorasi data serta pemeriksaan kesalahan atau ketidakkonsistenan.

Pada tahap ini, dilakukan serangkaian kegiatan yang meliputi pengumpulan data awal, pendeskripsian data, eksplorasi pola serta hubungan antar variabel, hingga verifikasi kualitas data sebelum masuk ke tahap pemodelan.

Sebagai contoh penerapan pada penelitian ini, proses Data Understanding dilakukan menggunakan Iris Dataset untuk memahami karakteristik data, distribusi kelas, serta hubungan antar atribut yang tersedia.
