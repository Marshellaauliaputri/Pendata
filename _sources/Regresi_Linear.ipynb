# Analisis Data Menggunakan Regresi Linier

## 1. Pengertian Regresi Linier

Regresi linier adalah metode analisis data yang digunakan untuk mengetahui hubungan antara variabel bebas $(x)$ dan variabel terikat $(y)$.

Metode ini bekerja dengan mencari garis lurus terbaik yang paling mendekati kumpulan data. Garis tersebut dapat digunakan untuk melihat pola hubungan antar data dan memprediksi nilai $(y)$ berdasarkan nilai $(x)$.

Bentuk umum regresi linier adalah:

$$
\hat{y} = b_0 + b_1x
$$

Keterangan:

- $(\hat{y})$ = nilai hasil prediksi
- $(x)$ = variabel bebas
- $(b_0)$ = intercept atau konstanta
- $(b_1)$ = koefisien regresi atau kemiringan garis

Pada analisis ini, regresi linier digunakan untuk mencari persamaan garis terbaik dari titik-titik data yang diperoleh dari GeoGebra.

## 3. Rumus Regresi Linier

Model regresi linier sederhana:

$$
\hat{y} = b_0 + b_1x
$$

Dalam bentuk matriks:

$$
Y = X\beta
$$

Untuk mencari vektor koefisien regresi \(\beta\), digunakan rumus normal equation:

$$
\hat{\beta} = (X^T X)^{-1}X^T Y
$$

Keterangan:

- $(X)$ = matriks fitur
- $(Y)$ = vektor target
- $(X^T)$ = transpose dari matriks X
- $((X^T X)^{-1})$ = invers dari \(X^T X\)
- $(\hat{\beta})$ = vektor koefisien regresi

Vektor koefisien regresi:


$$
\hat{\beta} =
\begin{bmatrix}
b_0 \\
b_1
\end{bmatrix}
$$

## 5. Menghitung Koefisien Regresi Menggunakan sklearn

Kode Python:




````python
import numpy as np
from sklearn.linear_model import LinearRegression

# Data dari grafik
X = np.array([[2], [3], [3], [3], [4], [4], [5]])
Y = np.array([2, 5, 4, 3, 3, 5, 6])

# Membuat model regresi linier
model = LinearRegression()

# Melatih model
model.fit(X, Y)

# Mengambil nilai intercept dan koefisien
intercept = model.intercept_
slope = model.coef_[0]

print("Intercept / b0 =", intercept)
print("Koefisien / b1 =", slope)

# Persamaan regresi
print(f"Persamaan regresi: y = {intercept:.2f} + {slope:.2f}x")


````

## 6. Program Menghitung β Secara Analitik

````python
import numpy as np

# Data
x = np.array([2, 3, 3, 3, 4, 4, 5])
y = np.array([2, 5, 4, 3, 3, 5, 6])

# Membuat matriks X dengan kolom pertama berisi 1 untuk intercept
X = np.column_stack((np.ones(len(x)), x))

# Mengubah y menjadi vektor kolom
Y = y.reshape(-1, 1)

# Menghitung beta menggunakan rumus normal equation
beta = np.linalg.inv(X.T @ X) @ X.T @ Y

print("Matriks X:")
print(X)

print("\nVektor Y:")
print(Y)

print("\nHasil beta:")
print(beta)

print("\nIntercept / b0 =", beta[0][0])
print("Koefisien / b1 =", beta[1][0])
````



## 7. Hasil Analisis

Dengan demikian:

- Koefisien regresi $(b_1 = 1.05)$

Persamaan regresi akhirnya adalah:

$
\hat{y} = 0.4 + 1.05x
$

Artinya, setiap nilai $(x)$ naik 1 satuan, maka nilai $(y)$ diprediksi naik sebesar 1.05 satuan.
