## Proses Data Mining dalam CRISP-DM

Proses data mining umumnya mengikuti metode CRISP-DM (Cross-Industry Standard Process for Data Mining), yaitu kerangka kerja yang membantu pelaksanaan proyek data mining secara terstruktur agar menghasilkan analisis yang sistematis dan sesuai kebutuhan.

CRISP-DM terdiri dari lima tahap utama, yaitu:

### 1. Business Understanding (Pemahaman Bisnis)

adalah tahap awal yang berfokus pada memahami masalah dan tujuan yang ingin dicapai. Pada tahap ini dilakukan identifikasi kebutuhan, penentuan tujuan proyek, serta perencanaan langkah kerja agar analisis memiliki arah yang jelas.

### 2. Data Understanding (Pemahaman Data)

merupakan tahap pengumpulan dan pengenalan data. Data dianalisis untuk mengetahui struktur, isi, dan kualitasnya melalui eksplorasi data serta pemeriksaan kesalahan atau ketidakkonsistenan.

Pada tahap ini, dilakukan serangkaian kegiatan yang meliputi pengumpulan data awal, pendeskripsian data, eksplorasi pola serta hubungan antar variabel, hingga verifikasi kualitas data sebelum masuk ke tahap pemodelan.

```python
# Statistik deskriptif

import pandas as pd
from scipy import stats

df = pd.read_csv("IRIS.csv")

# Ambil hanya kolom numerik (tanpa species)
df_numeric = df.select_dtypes(include=['int64','float64'])

for col in df_numeric.columns:
    print("===== Statistik untuk", col, "=====")
    print("Jumlah Data      :", df[col].count())
    print("Rata-rata        :", df[col].mean())
    print("Nilai Minimum    :", df[col].min())
    print("Q1               :", df[col].quantile(0.25))
    print("Median (Q2)      :", df[col].quantile(0.5))
    print("Q3               :", df[col].quantile(0.75))
    print("Nilai Maximum    :", df[col].max())
    print("Kemencengan      :", round(df[col].skew(), 6))

    mode = stats.mode(df[col], keepdims=True)
    print("Modus            :", mode.mode[0], "dengan jumlah", mode.count[0])

    print("Standar Deviasi  :", round(df[col].std(), 6))
    print("Variansi         :", round(df[col].var(), 6))
    print("\n")
```

##### **Menghitung Jarak pada Data dengan Tipe Campuran**

Data campuran merupakan data yang memiliki lebih dari satu jenis atribut dalam satu dataset, contohnya:

* Numerik → seperti panjang sepal (sepal length) dan lebar sepal (sepal width)

* Kategorikal → seperti jenis kelamin atau spesies bunga

* Ordinal → seperti peringkat atau tingkat kepuasan

✓ Apabila seluruh atribut bersifat numerik, maka perhitungan jarak dapat langsung menggunakan metode Euclidean Distance.

✓ Namun, jika dataset terdiri dari kombinasi beberapa tipe data, maka perhitungan jarak tidak bisa dilakukan secara langsung dengan metode tersebut.

Untuk menangani data bertipe campuran, dapat digunakan metode Gower Distance. Metode ini menghitung kontribusi setiap atribut sesuai dengan karakteristiknya masing-masing, lalu menghasilkan nilai jarak dalam rentang 0 hingga 1 sehingga lebih mudah untuk dianalisis dan diinterpretasikan.


##### **a. Rumus Gower Distance**


a. Perhitungan Rumus Grow distance

Untuk setiap variabel:

* Numerik :


$$
d_ij = \frac{|x_1 - x_j|}{Range}
$$


* Kategorikal:

$$
d_{ij} =
\begin{cases}
0, & \text{jika sama} \\
1, & \text{jika berbeda}
\end{cases}
$$


$$
D_{ij} = \frac{\sum d_{ij}}{\text{jumlah variabel}}
$$



Dimana:


- $D_{ij}$ = jarak antara objek $i$ dan $j$  


Nilai akhirnya berada di rentang:


$$
0 \leq D_{ij} \leq 1
$$


- 0 → sangat mirip
- 1 → sangat berbeda


**contoh:**


| Objek | Umur | IPK | Jenis Kelamin | Status Beasiswa |
| ----- | ---- | ----| ------------- | --------------- |
| A     | 20   | 3.5 | Perempuan     | Ya              |
| B     | 22   | 3.8 | Laki Laki     | Tidak           |
| C     | 21   | 3.2 | Perempuan     | Ya              |


Tipe datanya:

* Numerik → Umur, IPK

* Kategorikal → Jenis Kelamin, Status Beasiswa


1.  Hitung Range

Hitung Range Numerik

🔹 Umur

Range = 22 − 20 = 2

🔹 IPK

Range = 3.8 − 3.2 = 0.6

2. Hitung Jarak A dan B

🔹umur: $
|20 - 22| / 2 = 2 / 2 = 1
$
    
🔹IPK: $
|3.5 - 3.8| / 0.6 = 0.3 / 0.6 = 0.5
$

🔹 Jenis Kelamin: Perempuan ≠ Laki-laki → 1

🔹 Status Beasiswa: Ya ≠ Tidak → 1

Total:

$$
D(A, B) = (1 + 0.5 + 1 + 1) / 4
$$


$$
= 3.5 / 4
$$


$$
= 0.875
$$

3. Hitung jarak A dan C

🔹umur: $
|20 - 21| / 2 = 1 / 2 = 0.5
$
    
🔹IPK: $
|3.5 - 3.2| / 0.6 = 0.3 / 0.6 = 0.5
$

🔹 Jenis Kelamin: Perempuan= Perempuan → 0

🔹 Status Beasiswa: Ya = Ya → 0

Total:

$$
D(A, C) = (0.5 + 0.5 + 0 + 0) / 4
$$


$$
= 1 / 4
$$


$$
= 0.25
$$

4. Hitung jarak B dan C


🔹umur: $
|22 - 21| / 2 = 1 / 2 = 0.5
$
    
🔹IPK: $
|3.8 - 3.2| / 0.6 = 0.6 / 0.6 = 1
$

🔹 Jenis Kelamin: Laki Laki ≠ Perempuan → 1

🔹 Status Beasiswa: Tidak ≠ Ya → 1

Total:

$$
D(B, C) = (0.5 + 1 + 1 + 1) / 4
$$


$$
= 3.5 / 4
$$


$$
= 0.875
$$


Kesimpulan

* A paling mirip dengan C (jarak 0.25 kecil)

* B cukup jauh dari A dan C (0.875 besar)

* Semakin mendekati 0 → semakin mirip

* Semakin mendekati 1 → semakin berbeda


##### **b. Kelebihan Gower Distance**

* Mampu Mengolah Data Bertipe Campuran
Gower Distance dirancang untuk menghitung jarak pada dataset yang memiliki kombinasi atribut numerik, kategorikal, maupun ordinal dalam satu perhitungan. Hal ini menjadi keunggulan utama dibandingkan metode jarak lain yang umumnya hanya cocok untuk satu tipe data tertentu.

* Normalisasi pada Atribut Numerik
Setiap atribut numerik dinormalisasi menggunakan rentang (range) datanya. Dengan demikian, perbedaan skala antar variabel tidak menyebabkan satu atribut mendominasi hasil perhitungan jarak.

* Nilai Jarak Mudah Diinterpretasikan
Hasil perhitungan berada pada rentang 0 sampai 1, di mana nilai 0 menunjukkan kemiripan sempurna dan nilai 1 menunjukkan perbedaan maksimal. Rentang ini memudahkan interpretasi hasil analisis.

* Dapat Menangani Missing Value
Gower Distance tetap dapat menghitung jarak meskipun terdapat nilai yang hilang, dengan menyesuaikan jumlah atribut yang digunakan dalam perhitungan.

* Cocok untuk Analisis Clustering Data Heterogen
Metode ini sering digunakan dalam teknik clustering (misalnya hierarchical clustering) ketika data terdiri dari berbagai tipe atribut.

##### **c. Kekurangan Gower Distance**

* Perhitungan Relatif Lebih Kompleks
Karena setiap tipe atribut dihitung dengan cara berbeda (numerik, kategorikal, ordinal), proses komputasinya lebih kompleks dibandingkan metode jarak sederhana seperti Euclidean Distance.

* Dipengaruhi oleh Range Data Numerik
Jika terdapat outlier ekstrem, nilai range menjadi sangat besar sehingga dapat mengecilkan kontribusi perbedaan antar data pada atribut tersebut.

* Bobot Atribut Sama Secara Default
Secara standar, semua atribut memiliki kontribusi yang sama dalam perhitungan jarak. Jika ada atribut yang lebih penting, perlu dilakukan pembobotan secara manual.

* Kurang Efisien untuk Dataset Numerik Murni
Apabila seluruh atribut berupa numerik, metode seperti Euclidean Distance lebih sederhana dan lebih cepat digunakan.

* Komputasi Bisa Lebih Berat pada Dataset Besar
Karena menghitung kontribusi setiap atribut untuk setiap pasangan data, waktu komputasi dapat meningkat pada dataset dengan jumlah observasi yang sangat banyak.
